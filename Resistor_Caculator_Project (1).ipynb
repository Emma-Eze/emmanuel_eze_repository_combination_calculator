```python
# Resistor Combination Calculator
# Author: Emmanuel Eze
# This program calculates the equivalent resistance of resistors
# connected in series and in parallel.


def get_positive_number(prompt):
    """Ask for a positive number and return it as a float."""
    while True:
        try:
            value = float(input(prompt))
            if value <= 0:
                print("Please enter a value greater than 0.")
            else:
                return value
        except ValueError:
            print("Invalid input. Please enter a number, for example 220 or 4.7.")


def get_resistor_count():
    """Ask how many resistors will be entered and return the number."""
    while True:
        try:
            count = int(input("How many resistors do you have? "))
            if count < 2:
                print("Please enter at least 2 resistors.")
            else:
                return count
        except ValueError:
            print("Invalid input. Please enter a whole number, for example 3.")


def calculate_series(resistors):
    """Return the total resistance for a series connection."""
    return sum(resistors)


def calculate_parallel(resistors):
    """Return the total resistance for a parallel connection."""
    reciprocal_sum = 0
    for resistance in resistors:
        reciprocal_sum += 1 / resistance
    return 1 / reciprocal_sum


def format_resistance(value):
    """Return resistance in a convenient unit: ohms, kOhms or MOhms."""
    if value >= 1_000_000:
        return f"{value / 1_000_000:.3f} MOhms"
    elif value >= 1_000:
        return f"{value / 1_000:.3f} kOhms"
    else:
        return f"{value:.3f} ohms"


def main():
    print("=" * 50)
    print("        RESISTOR COMBINATION CALCULATOR")
    print("=" * 50)
    print("Enter all resistor values in ohms.\n")

    count = get_resistor_count()

    # A list stores all resistor values entered by the user.
    resistors = []
    for number in range(1, count + 1):
        resistance = get_positive_number(f"Enter resistor {number} value (ohms): ")
        resistors.append(resistance)

    series_total = calculate_series(resistors)
    parallel_total = calculate_parallel(resistors)

    print("\n--- RESULTS ---")
    print(f"Resistor values: {resistors}")
    print(f"Series total:   {format_resistance(series_total)}")
    print(f"Parallel total: {format_resistance(parallel_total)}")

    # Own twist: compare the two results and show the difference.
    difference = series_total - parallel_total
    print(f"Difference:     {format_resistance(difference)}")

    if parallel_total < min(resistors):
        print("Check: The parallel total is correctly less than the smallest resistor.")
    elif parallel_total == min(resistors):
        print("Check: The parallel total equals the smallest resistor.")
    else:
        print("Check: Please review the resistor values entered.")

    print("\nThank you for using the calculator.")


# Start the program.
if __name__ == "__main__":
    main()
```
